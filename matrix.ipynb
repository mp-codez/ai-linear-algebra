```python
import numpy as np
A = np.array([[1, 3, 4],[6, 4, 1],[2, 2, 2]])
A
```

```python
B = np.array([5, 16, 11])
B
```

```python
A_inv = np.linalg.inv(A)
A_inv
```

```python
x, y, z = np.dot(A_inv, B)
```
